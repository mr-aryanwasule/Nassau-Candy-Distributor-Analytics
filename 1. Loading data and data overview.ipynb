```python
# Import required libraries
import pandas as pd
import numpy as np
```

```python
# Dataset path
file_path = r"Nassau_Candy_Distributor.csv"

# Load dataset
df = pd.read_csv(file_path)
```

```python
# Display first 5 rows
print("First 5 Rows:")
print(df.head())
```

```python
# Display the last 5 rows
print("\nLast 5 Rows:")
print(df.tail())
```

```python
# Check data set shape
print("\nDataset Shape:")
print(df.shape)
```

```python
# Display column names
print("\nColumn Names:")
print(df.columns)
```

```python
# Data Set Information
print("\nDataset Information:")
df.info()
```

```python
# Statical Summary
print("\nStatistical Summary:")
print(df.describe())
```

```python
# Check data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Check Duplicate Rows
duplicates = df.duplicated().sum()
```

```python
print("Duplicate Rows:", duplicates)
# Check unique value of each column
print("\nUnique Values in Each Column:")
```

```python
for column in df.columns:
    print(f"{column}: {df[column].nunique()}")
```

```python
# Create a data set summary table
summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum(),
    "Unique Values": df.nunique()
})

print(summary)
```

```python
# Save the summary
summary.to_csv("dataset_summary.csv")

print("Dataset summary saved successfully.")
```
