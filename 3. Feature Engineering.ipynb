```python
# Import Libraries
import pandas as pd
import numpy as np
```

```python
# Load the Cleaned Dataset
file_path = r"cleaned_nassau_candy.csv"

df = pd.read_csv(file_path)
```

```python
# Convert Date Columns
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
# Create Year Feature
df["Order Year"] = df["Order Date"].dt.year
```

```python
# Create Month Feature
df["Order Month"] = df["Order Date"].dt.month
```

```python
# Create Month Name
df["Month Name"] = df["Order Date"].dt.month_name()
```

```python
# Create Quarter
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
# Create Day Feature
df["Order Day"] = df["Order Date"].dt.day
```

```python
# Create Day of Week
df["Day of Week"] = df["Order Date"].dt.day_name()
```

```python
# Weekend Indicator
df["Is Weekend"] = df["Order Date"].dt.dayofweek >= 5
```

```python
# Shipping Time
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
# Sales per Unit
df["Sales Per Unit"] = (
    df["Sales"] / df["Units"]
).round(2)
```

```python
# Cost per Unit
df["Cost Per Unit"] = (
    df["Cost"] / df["Units"]
).round(2)
```

```python
# Profit per Unit
df["Profit Per Unit"] = (
    df["Gross Profit"] / df["Units"]
).round(2)
```

```python
# Profit Margin (%)
df["Profit Margin (%)"] = (
    (df["Gross Profit"] / df["Sales"]) * 100
).round(2)
```

```python
# Shipping Efficiency
df["Shipping Efficiency"] = np.where(
    df["Shipping Days"] <= 2,
    "Fast",
    np.where(
        df["Shipping Days"] <= 5,
        "Normal",
        "Slow"
    )
)
```

```python
# Sales Category
df["Sales Category"] = pd.cut(
    df["Sales"],
    bins=[0,10,25,50,100,1000],
    labels=[
        "Very Low",
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
# Profit Category
df["Profit Category"] = pd.cut(
    df["Gross Profit"],
    bins=[-1,5,10,20,40,500],
    labels=[
        "Very Low",
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
# Season
def season(month):
    if month in [12,1,2]:
        return "Winter"
    elif month in [3,4,5]:
        return "Summer"
    elif month in [6,7,8,9]:
        return "Monsoon"
    else:
        return "Autumn"

df["Season"] = df["Order Month"].apply(season)
```

```python
# High Demand Flag
average_sales = df["Sales"].mean()

df["High Demand"] = df["Sales"] > average_sales
```

```python
# Total Revenue Check
df["Revenue Check"] = (
    df["Cost"] + df["Gross Profit"]
).round(2)
```

```python
# Display New Features
print(df.head())
```

```python
# Check New Dataset Information
print(df.info())
```

```python
 # Save the Feature Engineered Dataset
output_path = r"feature_engineered_nassau_candy.csv"

df.to_csv(output_path, index=False)

print("Feature engineered dataset saved successfully.")
```
