```python
# Import Libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
file_path = r"Nassau_Candy_Distributor.csv"

df = pd.read_csv(file_path)
```

```python
# Display Dataset Information
print(df.info())
```

```python
# Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Remove extra spaces from all text columns
text_columns = df.select_dtypes(include=["object", "string"]).columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("Extra spaces removed successfully.")
```

```python
# Convert Date Columns
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True,
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True,
    errors="coerce"
)

print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
# Remove Invalid Date Records
df = df.dropna(subset=["Order Date", "Ship Date"])
# Convert Numerical Columns
numeric_columns = [
    "Sales",
    "Units",
    "Gross Profit",
    "Cost"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Remove Invalid Numeric Records
df.dropna(subset=numeric_columns, inplace=True)
```

```python
# Check Negative Values
print("Negative Sales:", (df["Sales"] < 0).sum())

print("Negative Cost:", (df["Cost"] < 0).sum())

print("Negative Profit:", (df["Gross Profit"] < 0).sum())
```

```python
# Verify Data Types
print(df.dtypes)
```

```python
# Check Dataset Shape
print("Shape:", df.shape)
```

```python
# Final Missing Values Check
print(df.isnull().sum())
```

```python
# Save the Cleaned Dataset

output_path = r"cleaned_nassau_candy.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
```

```python
# Verify the Cleaned Dataset
clean_df = pd.read_csv(output_path)

print(clean_df.head())

print(clean_df.info())

print(clean_df.shape)
```
